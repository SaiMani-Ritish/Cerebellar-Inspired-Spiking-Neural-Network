# Smoke test: `evaluate.py`

```python
import sys
import numpy as np

# Ensure the `snn_research` package is importable regardless of nbconvert CWD.
sys.path.insert(0, r"C:\Users\sriha\Downloads\SNN")

from snn_research.evaluate import compute_jerk_series, oscillation_amplitude, log_episode

# Jerk test: linear ramp should have small 3rd derivative
positions = np.linspace(0, 1, 200, dtype=np.float64)
peak_jerk = compute_jerk_series(positions, dt=0.02)
print("peak_jerk(linear):", peak_jerk)

# Oscillation amplitude: zeros then sinusoid tail
tail_t = np.arange(100) * 0.02
tail = 0.1 * np.sin(2 * np.pi * 2 * tail_t)  # RMS ~ 0.1/sqrt(2)
positions2 = np.concatenate([np.zeros(100), tail])
osc_amp = oscillation_amplitude(positions2, stop_idx=100, window=100)
print("osc_amp(sinus tail):", osc_amp)

metrics = log_episode(
    controller_name="smoke",
    positions=[0.0, 0.5, 0.2],
    spike_count=123,
    flop_count=456,
)
print("log_episode keys:", sorted(metrics.keys()))
print("log_episode sample:", metrics)

print("EVALUATE SMOKE TEST PASSED")
```
